```python
#-------------------------------------------
# Disease Prediction from Medical Data
#-------------------------------------------

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.datasets import load_breast_cancer


#-------------------------------------------
# 1. Load Datasets (working URLs)
#-------------------------------------------

url_heart = "https://raw.githubusercontent.com/ryanleeallred/datasets/master/heart.csv"
url_diabetes = "https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv"

heart_df = pd.read_csv(url_heart)
diabetes_df = pd.read_csv(url_diabetes)

# Breast Cancer Dataset (built-in, no errors ever)
cancer = load_breast_cancer()
cancer_df = pd.DataFrame(cancer.data, columns=cancer.feature_names)
cancer_df['target'] = cancer.target

print("Datasets Loaded Successfully!")


#-------------------------------------------
# 2. Preprocessing Function
#-------------------------------------------

def preprocess(df, target_column):
    X = df.drop(columns=[target_column])
    y = df[target_column]

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    return train_test_split(X_scaled, y, test_size=0.2, random_state=42)


#-------------------------------------------
# 3. Machine Learning Models
#-------------------------------------------

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "SVM": SVC(),
    "Random Forest": RandomForestClassifier(),
    "XGBoost": XGBClassifier(eval_metric="logloss")
}


#-------------------------------------------
# 4. HEART DISEASE PREDICTION
#-------------------------------------------

print("\n----- HEART DISEASE PREDICTION -----")
X_train, X_test, y_train, y_test = preprocess(heart_df, "target")

for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    acc = accuracy_score(y_test, preds)
    print(f"{name}: {acc:.4f}")


#-------------------------------------------
# 5. DIABETES PREDICTION
#-------------------------------------------

print("\n----- DIABETES PREDICTION -----")
X_train, X_test, y_train, y_test = preprocess(diabetes_df, "Outcome")

for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    acc = accuracy_score(y_test, preds)
    print(f"{name}: {acc:.4f}")


#-------------------------------------------
# 6. BREAST CANCER PREDICTION
#-------------------------------------------

print("\n----- BREAST CANCER PREDICTION -----")
X_train, X_test, y_train, y_test = preprocess(cancer_df, "target")

for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    acc = accuracy_score(y_test, preds)
    print(f"{name}: {acc:.4f}")


```
